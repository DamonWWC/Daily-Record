```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.DataFrame({
    'year': [2010, 2011, 2012, 2010, 2011, 2012, 2010, 2011, 2012],
    'sales': [54, 48, 59, 41, 44, 51, 40, 44, 50],
})
```

```python
plt.bar(df['year'],df['sales'],color='blue')
plt.xlabel('year')
plt.ylabel('sales')
plt.title('year vs sales')
plt.show()
```
