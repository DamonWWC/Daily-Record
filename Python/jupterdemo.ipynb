```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")
say_hello()

def my_decorator1(func):
    def wrapper(*args,**kwargs):
        print("Something is happening before the function is called.")
        func(*args,**kwargs)
        print("Something is happening after the function is called.")
    return wrapper
@my_decorator1
def greet(name):
    print(f"Hello {name}!")
greet("Runoob")

def repeat(num_times):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(num_times):
                func(*args,**kwargs)
        return wrapper
    return decorator

@repeat(4)
def say_hello():
    print("Hello!")
say_hello()


def my_decorator2(cls):
    class Wrapper:
        def __init__(self,*args,**kwargs):
            self.wrapped=cls(*args,**kwargs)
        def display(self):
            print("Something is happening before the function is called.")
            self.wrapped.display()
            print("Something is happening after the function is called.")
    return Wrapper

@my_decorator2
class MyClass:
    def display(self):
        print("Hello World!")
obj=MyClass()
obj.display()
```

```python
class MyClass:
    @staticmethod #  将方法定义为静态方法，不需要实例化类即可调用。
    def static_method():  # 定义一个名为static_method的函数
        print("This is a static method.")  # 在函数内部打印字符串"This is a static method."
    @classmethod #  将方法定义为类方法，需要实例化类即可调用。
    def class_method(cls):  # 定义一个名为class_method的函数，cls表示类本身
        print("This is a class method.")  # 在函数内部打印字符串"This is a class method."
    @property
    def name(self):  # 定义一个名为instance_method的函数，self表示实例对象
        return self._name  # 在函数内部打印字符串"This is an instance method."
    
    @name.setter
    def name(self, value):  # 定义一个名为name的属性，并设置setter方法
        self._name = value  # 将传入的值赋值给实例对象的_name属性

MyClass.static_method()  # 调用静态方法
MyClass.class_method()  # 调用类方法
obj = MyClass()  # 实例化类
obj.name = "Runoob"  # 调用setter方法设置属性值
print(obj.name)  # 调用属性
```

```python
#生产者-消费者模型（Condition）
import random
from threading import Condition

queue= []
cond=Condition()
MAX_ITEMS=5

def producer():
    for _ in range(10):
        with cond:
            while len(queue)>=MAX_ITEMS:
                cond.wait()
            item=random.randint(1,100)
            queue.append(item)
            print(f"Produced{item}")
            cond.notify()
def consumer():
    for _ in range(10):
        with cond:
            while not queue:
                cond.wait()
            item=queue.pop(0)
            print(f"Consumed{item}")
            cond.notify()
threading.Thread(target=producer).start()
threading.Thread(target=consumer).start()
```

```python
import asyncio
import nest_asyncio
nest_asyncio.apply()

async def task(semaphore,task_id):
    async with semaphore:
        print(f"Task {task_id} is running")
        await asyncio.sleep(1)
        print(f"Task {task_id} is done")
async def main():
    semaphore=asyncio.Semaphore(3)
    tasks=[task(semaphore,i) for i in range(10)]
    await asyncio.gather(*tasks)
asyncio.run(main())
```
